```python
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import Sequential,optimizers
from tensorflow.keras.layers import Dense
from tensorflow.keras.datasets import mnist

# Load MNIST
(x,y),(xt,yt)=mnist.load_data()
x=x.reshape(-1,784).astype("float32")/255
xt=xt.reshape(-1,784).astype("float32")/255

# Model
def M(opt):
    m=Sequential([
        Dense(64,activation="relu",input_shape=(784,)),
        Dense(32,activation="relu"),
        Dense(10,activation="softmax")
    ])
    m.compile(optimizer=opt,
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
    return m

# Optimizers
ops={
    "GD":optimizers.SGD(.01),
    "Momentum":optimizers.SGD(.01,momentum=.9),
    "RMSProp":optimizers.RMSprop(.001),
    "Adam":optimizers.Adam(.001)
}

results={}

# Training
for name,opt in ops.items():
    print("\nTraining:",name)
    m=M(opt)
    h=m.fit(x,y,epochs=10,validation_split=.2,verbose=0)
    acc=m.evaluate(xt,yt,verbose=0)[1]
    results[name]=(h,acc)
    print("Test Accuracy:",round(acc,4))

# Accuracy graph
for name,(h,acc) in results.items():
    plt.plot(h.history["accuracy"],label=name)
plt.title("Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.show()

# Loss graph
for name,(h,acc) in results.items():
    plt.plot(h.history["loss"],label=name)
plt.title("Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.show()

# Final comparison
print("\nPerformance Comparison")
for name,(h,acc) in results.items():
    print(name,"-> Accuracy:",round(acc,4),
          "Loss:",round(h.history["loss"][-1],4))
```
